```python
activity_df = pd.read_csv(const.PATH_TO_RAW_ACTIVITY_DATA)
vehicle_df = pd.read_csv(const.PATH_TO_RAW_VEHICLE_DATA)
```

```python
activity_id_na = activity_df["vehicle_id"].isna().sum()
vehicle_id_na = vehicle_df["vehicle_id"].isna().sum()

print("NaN vehicle_id count in activity.csv:", activity_id_na)
print("NaN vehicle_id count in vehicle.csv:", vehicle_id_na)
```

```python
activity_ids = set(activity_df["vehicle_id"].dropna().unique())
vehicle_ids = set(vehicle_df["vehicle_id"].dropna().unique())

# IDs in activity but not in vehicle
missing_in_vehicle = activity_ids - vehicle_ids

# IDs in vehicle but not in activity
missing_in_activity = vehicle_ids - activity_ids

print("All activity vehicle_ids exist in vehicle_df:", len(missing_in_vehicle) == 0)
print("All vehicle_df vehicle_ids exist in activity_df:", len(missing_in_activity) == 0)
print()
print("Missing in vehicle_df:", len(missing_in_vehicle))
print("Missing in activity_df:", len(missing_in_activity))
```
